## Anaconda vs. VS Code

Anaconda and Visual Studio Code serve different roles and can be used together. Anaconda bundles Python, hundreds of packages, and environment-management tools, making it convenient but requiring more disk space. VS Code is a lightweight code editor that connects to your Python installation and lets you add extensions as needed. The table below compares their purposes and key features.

::: {.tools-comparison}

| | ![](images/anaconda-icon.svg){.comparison-logo fig-alt=""} **Anaconda** | ![](images/vscode-icon.svg){.comparison-logo fig-alt=""} **Visual Studio Code (VS Code)** |
|---|---|---|
| **Overview** | A Python distribution with environment-management tools | A powerful, flexible, and lightweight code editor |
| **What it is** | A distribution that includes Python, many common packages, and tools such as conda and Navigator | A free code editor from Microsoft, built on the open-source Code - OSS project |
| **Main purpose** | Makes it easy to install Python and manage packages and environments | Provides a modern, flexible environment to write, run, and debug code in Python and many other languages |
| **Key features** | • Preinstalls many popular data science packages<br><br>• Environment management with conda<br><br>• Anaconda Navigator graphical interface | • Clean, customizable interface<br><br>• Python support through extensions<br><br>• Integrated terminal and Jupyter notebook support through the Jupyter extension<br><br>• Debugging, code navigation, and Git tools<br><br>• Extensible through additional extensions<br><br>• Works with different Python environments, including Anaconda environments |

:::

### Why We Use VS Code Throughout STAT303

Throughout the STAT303 Python sequence, you will move from running individual code cells to managing complete analyses: reading data, writing and checking code, explaining results, and preparing reports. **VS Code gives you one workspace for that process.** Learning the workflow now lets you reuse it as the assignments become more complex.

- **Build on what you learned in STAT 201.** You will still use Jupyter notebooks (`.ipynb` files), with code cells, Markdown explanations, and outputs. VS Code becomes the place where you edit and run those notebooks, and it can connect to the Python installation you already have through Anaconda.
- **Keep each analysis organized.** Opening a project folder puts your notebooks, datasets, and supporting files together in the Explorer. You can see which files belong to an assignment and practice using paths that work within that project.
- **Understand and troubleshoot your code.** Code completion and function information help you work with unfamiliar libraries. Seeing code, outputs, and error messages together makes it easier to investigate a problem and check whether a change fixes it.
- **Use the same workflow for each submission.** Run and check your notebook, save it, render it to HTML with Quarto from the integrated terminal, and inspect the report before submitting. Practicing these steps makes the submission process familiar across the sequence.

The tools have distinct jobs: **VS Code is the workspace, the Python kernel runs your notebook code, and Quarto creates the HTML report.** The notebook kernel and the terminal can use different Python environments; in this chapter and the next, you will learn how to check that each uses the environment you intend.

## Code, Markdown, and Saved Outputs

Use code cells for computation and Markdown cells for explanations, headings, and answers. Run cells with their play controls or `Shift+Enter`. Save the notebook with `Ctrl+S` or `Cmd+S`.

A notebook remembers values from previously executed cells. Moving or editing a cell does not undo its earlier effects. Before considering work complete, **restart the kernel, run all code cells in order, and save again**. A successfully executed notebook should not depend on variables created in deleted cells.

The HTML submitted later will be generated from this saved notebook. Writing an explanation in a code comment is useful, but Markdown is better for the report's narrative.


If your notebook is working, continue to the next section. If you encounter a problem, expand the troubleshooting guide below.

## Render and Submit with Quarto

In STAT201, you already used Quarto to generate HTML from notebooks in your Anaconda-based workflow. The same Quarto application works with notebooks edited in VS Code. This is a refresher on completing that workflow in the new editor.

### Check Your Existing Quarto Installation

Open **Terminal → New Terminal** in VS Code and enter:

```text
quarto --version
```

A version number means this terminal can find Quarto. You do not need to reinstall it simply because you changed editors. If the command is not found, restart VS Code and check your existing Quarto installation. If it is missing, install the **Quarto application** from the [official download page](https://quarto.org/docs/get-started/) and open a new terminal.

The **Quarto VS Code extension is optional** for our workflow. It adds editing assistance and integrated previews; it is not required to run `quarto render` in the terminal. The Quarto application, the Python/Jupyter extensions used for notebooks, and the optional Quarto extension serve different purposes. See [Quarto's VS Code guide](https://quarto.org/docs/tools/vscode/index.html).

### Keep the Template's Report Settings

The practice notebooks start with a **Raw** cell containing settings like these:

```yaml
---
title: "Activity 1: Your First VS Code Notebook"
author: "Your Name"
format:
  html:
    embed-resources: true
    toc: true
    code-fold: false
---
```

Edit the existing title and author when appropriate; retain the indentation and `---` lines. This is YAML configuration, not Python code. Keep it in the supplied Raw cell. `toc: true` adds a table of contents, and `code-fold: false` keeps code visible for grading.

`embed-resources: true` puts ordinary figures and styling inside the HTML. It does not include your Python environment or input data for rerunning the work, and external websites or services can still require internet access.

### Restart, Run All, Save, Render, Inspect

Quarto normally renders an `.ipynb` file using its **saved outputs**. It does not rerun notebook code by default. A successful render alone does not show that your current code runs correctly. Follow this sequence:

1. Select the intended notebook kernel. For Chapter 1, use your familiar working STAT201 environment; Chapter 2 introduces a project environment.
2. Restart the kernel and run every code cell from top to bottom. Resolve unexpected errors. Use Markdown headings and explanations near the results they discuss.
3. Save the notebook. Keep necessary imports and computations, and remove irrelevant exploratory output.
4. In the terminal, check that you are in the folder containing the notebook. List its files with `ls` on macOS, `Get-ChildItem` in Windows PowerShell, or `dir` in Windows Command Prompt. If needed, use `cd` with the actual folder path in quotes; Chapter 2 explains navigation in detail.
5. Run `quarto render filename.ipynb --to html`, replacing `filename.ipynb` with your notebook's name. Quote a filename containing spaces, for example `quarto render "my assignment.ipynb" --to html`.
6. Open the resulting HTML in a browser. Check your name or required identifier, explanations, code, and current outputs. Copy **only the HTML** to another folder and open the copy to check that ordinary figures and formatting travel with it. Correct the notebook and repeat saving/rendering if anything is missing.

The standalone practice kits have no Quarto project configuration redirecting output, so the HTML is created beside the notebook. Keep the notebook and data for future work; upload the HTML requested in the activity or assignment. Detailed notebook rendering behavior is described in [Quarto's notebook guide](https://quarto.org/docs/tools/jupyter-lab.html).

::: {.callout-note collapse="true" title="Troubleshooting rendering and saved output"}

| Symptom | First check |
|---|---|
| Quarto command not found | Existing Quarto application installation and a fresh VS Code terminal |
| No valid input files | Terminal folder and the notebook's exact filename |
| Missing or old outputs in HTML | Restart/run all, save, render again, and open the newly generated HTML |
| Raw YAML appears in the report | Keep the settings in the supplied Raw cell rather than a Markdown or code cell |
| Figure or styling missing from the copied HTML | Confirm `embed-resources: true` in the Raw cell and render again |
| Notebook import or file error | Resolve it before rendering; Chapter 2 covers environment and file-path checks |

:::
