## 4. Exercise Minutes (16 points) {-}

*An application of vectorized random sampling.*

The list `exercise_minutes`, supplied in the starter, holds the weekly exercise minutes reported by students in a past STAT 303-1 class. This question finds a 95% confidence interval for the mean with **bootstrapping**, twice — once in plain Python and once with NumPy — and compares the time each takes.

The bootstrap method:

1. Let $N$ be the number of values in `exercise_minutes`.
2. Draw $N$ values at random, **with replacement**, from `exercise_minutes`.
3. Compute the mean of those $N$ values.
4. Repeat steps 2 and 3 10,000 times.
5. The 95% confidence interval runs from the 2.5th to the 97.5th percentile of the 10,000 means.

The starter notebook supplies `exercise_minutes` as a Python list of 191 values.

**Hints:** `random.choices(exercise_minutes, k=N)`, from Python's built-in [`random` module](https://docs.python.org/3/library/random.html#random.choices) (`import random`), draws $N$ values with replacement. With the 10,000 means in a sorted list, the 2.5th and 97.5th percentiles are approximately the values at positions 250 and 9,750. Time the cell with `%timeit -o -r 3 -n 1`, as in [NumPy and pandas in a Real Workflow](numpy_pandas_workflow.ipynb), and keep the result's `.average`.

### 4.2 With NumPy (7 points) {-}

Using NumPy and **no loops**, compute the same confidence interval and time it the same way.

**Hint:** `rng = np.random.default_rng(303)`, then [`rng.choice()`](https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.choice.html), which draws with replacement by default, as in [NumPy Fundamentals](numpy_fundamentals.ipynb). Passing a tuple, `size=(10_000, N)`, returns a 2D array with one bootstrap sample per row, so `.mean(axis=1)` gives all 10,000 means at once.
