## What Is a Terminal?

A **terminal** is a text interface where you type commands and read their output. Instead of clicking a folder or a button, you describe an action in text. A **shell** is the program that reads those commands and carries them out. Common shells include zsh or Bash on macOS, and PowerShell or Command Prompt on Windows.

If you typed a command such as `quarto render assignment.ipynb --to html` in STAT 201, you have already used a terminal: you asked the shell to start Quarto, gave it a notebook filename, and requested HTML output. You do not need to run this example now.

In VS Code, open **Terminal → New Terminal**. This integrated terminal lets you work beside your notebook without opening a separate application. Check its tab or dropdown for the shell name; this chapter provides separate commands where shells differ. See the [VS Code terminal introduction](https://code.visualstudio.com/docs/terminal/getting-started) for more about the interface.

## Ask Three Questions Before a Command

Throughout this course, you will use both notebooks and the terminal. Some tasks can be completed in either place, but the commands may differ. Your notebook and terminal may also use different Python environments or work from different folders.

Before running a command, ask these three questions to help avoid syntax errors, installing packages in the wrong environment, or failing to find files.

| Question | In a notebook | In a terminal |
|---|---|---|
| **1. Where am I typing?** | Use a Python code cell for Python code. | Identify the shell; a `>>>` prompt means you are inside Python instead. |
| **2. Which Python am I using?** | Run `import sys`, then `print(sys.executable)`. | Run `python -c "import sys; print(sys.executable)"`. |
| **3. Which folder is current?** | Run `from pathlib import Path`, then `print(Path.cwd())`. | Use `pwd`, `Get-Location`, or `cd` for your shell. |

The terminal command `python -c "import sys; print(sys.executable)"` starts Python, prints the executable path, and exits back to the shell. On macOS, use `python3` instead if `python` is not available. Navigation commands such as `cd` do not use Python, so you do not need to check which Python environment is active before running them.

The notebook and terminal can use **different Python installations and different working directories**. Choosing a notebook kernel does not necessarily change which Python the terminal uses. Compare the two `sys.executable` outputs to identify which Python each uses; at this point, simply record what you find.

Similarly, the folder visible in Explorer does not prove either one's current directory. Check the current directory in each place. A terminal command may find or create a file in one folder while the notebook looks for it in another.

## Why Use a Python Virtual Environment?

In Chapter 1, you selected Anaconda’s **`base` environment** from STAT 201 to check that your notebook could run in VS Code. That gave you a familiar starting point. As you begin new projects, use a **separate environment for each project** instead of installing everything into `base`.

### Different Projects Need Different Tools

Your computer may have **multiple Python versions** installed to support different projects. One project might require Python 3.11, while another uses Python 3.12. These are examples, not instructions to install both versions for this course. Code and packages that work with one Python version may need changes to work with another.

Projects also use different **libraries**, installed as packages. For example, Pandas helps analyze tables, while Matplotlib helps create plots. Each package has its own versions, separate from Python's version, and may depend on other packages. These required packages are called **dependencies**.

Conflicts can arise when two tools need incompatible versions of the same dependency. As a hypothetical example, an older analysis tool might require version 1 of a shared library, while a newer tool requires version 2. Installing or upgrading packages for one project in a shared environment can make another project's previously working code fail.

### Give Each Project Its Own Environment

A **Python virtual environment** gives a project access to a particular Python interpreter and its own set of installed packages. Think of it as a toolbox assigned to that project. Two projects can use different Python versions, different packages, or different versions of the same package. They can also use the same Python version while keeping their packages separate.

![Each project uses its own Python environment. The versions and shared library shown are hypothetical examples, not course installation requirements.](images/python-project-environments.svg){#fig-python-project-environments width=100% fig-alt="Two projects on one computer use separate environments. Project A uses Python 3.11 and shared library version 1; Project B uses Python 3.12 and shared library version 2. Each notebook connects to its own environment, so changing packages in one does not change the other. An existing Python installation is shown as the Chapter 1 starting point."}

With separate environments, changing packages in Project B does not change the packages installed in Project A. Isolation does not make incompatible libraries work together inside one environment; each project still needs a compatible set of dependencies.

**For STAT303:** keep the notebooks in `stat303-setup` together as one course project and give that project one environment. You do not need a new environment for every notebook. ## Notebook Magic Commands

A **magic command** is an instruction to the notebook itself rather than a line of Python. These commands come from IPython, the kernel that runs your notebook code cells, and they exist to do things the Python language has no syntax for: install a package into the running environment, measure how long code takes, or configure how plots are displayed.

You have already run one. In `%pip install seaborn`, the leading `%` marks that line as a magic command rather than as Python.

This section introduces only the handful of magic commands that appear in later chapters. You do not need to memorize them now. The goal is that a line beginning with `%` looks familiar, rather than puzzling, when you meet one later.

### A Single `%` Marks a Line, and `%%` Marks a Cell

One percent sign marks a **line magic**, which applies to the rest of that one line only; the other lines in the cell stay ordinary Python. Two percent signs mark a **cell magic**, which applies to the entire cell and must be written as the cell's first line.

```python
%pwd
```

```python
%%timeit
total = 0
for n in range(100000):
    total = total + n
```

As noted for `%pip` above, automagic lets you drop the `%` from many line magics. Type it anyway. The symbol tells a reader, including you in a few weeks, that the line is an instruction to the notebook and not Python code.

### `%timeit` Measures How Long Code Takes to Run

`%timeit` reports how long a piece of code takes. Later chapters use it whenever two ways of writing the same calculation need to be compared. It runs the code repeatedly and reports an average, because a single run of fast code measures mostly noise.

Use `%timeit` for one line and `%%timeit` for a whole cell:

```python
%timeit sum(range(100000))
```

The result looks like the following. Your numbers will differ from a classmate's, because timing depends on the machine.

```text
1.12 ms ± 18.3 µs per loop (mean ± std. dev. of 7 runs, 1,000 loops each)
```

Read that as: the code was run 1,000 times in a row to form one loop's measurement, this was repeated 7 times, and one execution averaged 1.12 milliseconds. A millisecond (`ms`) is a thousandth of a second and a microsecond (`µs`) is a millionth.

`%time` runs the code once and reports that single run. Use it when one run already takes long enough to be informative, such as reading a large file, and `%timeit` when the code is fast.

::: {.callout-warning title="A `%%timeit` cell discards its variables"}
Because `%%timeit` runs the cell many times, any variable assigned inside it is discarded once timing finishes. Build your data in one cell, then time the calculation in another cell that only reads it.
:::

In [NumPy and pandas in a Real Workflow](numpy_pandas_workflow.ipynb), you will use `%timeit` to compare a row-by-row `.apply()` against the equivalent NumPy array calculation and see how much time the array version saves.

### `%matplotlib inline` Displays Plots Inside the Notebook

`%matplotlib inline` tells Matplotlib to draw each figure as a static image in the notebook's output area, directly beneath the cell that created it, instead of opening it in a separate window.

You will find this line at the top of many notebooks, including several later chapters of this book. VS Code's notebook editor already displays figures inline, so your plots appear whether or not you run it. Leave the line in place when you find it: it is harmless, and it makes the notebook behave the same way for someone who opens it in a different tool.

### A Short Reference

| Magic command | What it does |
|---|---|
| `%pip install <package>` | Installs a package into the environment running this notebook |
| `%timeit <code>` | Times one line, repeated, and reports the average |
| `%%timeit` | Times the whole cell, repeated; must be the cell's first line |
| `%time <code>` | Times one line, run once |
| `%matplotlib inline` | Draws Matplotlib figures in the notebook output |
| `%pwd` | Prints the notebook's current working directory |
| `%cd <folder>` | Changes the working directory for the rest of the kernel session |
| `%whos` | Lists the variables currently defined in the kernel |
| `%lsmagic` | Lists every available magic command |

A related symbol is `!`, which sends the rest of the line to your shell, as in `!ls`. This book prefers `%pip install` over `!pip install` because `%pip` is guaranteed to install into the environment your notebook kernel is using, while `!pip` runs whichever `pip` the shell happens to find.

To explore further, run `%lsmagic` for the full list, or add a trailing question mark, as in `%timeit?`, to read the built-in help for any magic command. The complete reference is IPython's [built-in magic commands](https://ipython.readthedocs.io/en/stable/interactive/magics.html).

### Magic Commands Work in Notebooks, Not in Scripts or the Terminal

Magic commands are IPython syntax, not Python language syntax. Copy `%timeit sum(range(100000))` into a `.py` file and run it with `python script.py`, and Python stops with a `SyntaxError`. They are also distinct from terminal commands: `pip install seaborn` is what you type at a shell prompt, while `%pip install seaborn` is what you type in a notebook code cell.

This is the first of the three questions from Section 2.4 in a new form. Before running a line that starts with `%`, confirm that you are typing it in a notebook code cell.

Here `>` writes output into a file. In PowerShell, `|` passes output to `Out-File`, and `-Encoding utf8` specifies the text encoding. This avoids the different default used by Windows PowerShell 5.1. See [PowerShell’s encoding reference](https://learn.microsoft.com/en-us/powershell/module/microsoft.powershell.core/about/about_character_encoding) and [pip’s requirements-file encoding](https://pip.pypa.io/en/stable/reference/requirements-file-format/#encoding).

All three commands write `requirements.txt` in the current folder and **replace it if it already exists**. The command may finish without displaying output; open the file in Explorer and check that it contains package entries before continuing.

A line such as `numpy==2.2.0` means “install NumPy version 2.2.0.” This is an illustration, not a required course version. Keep the actual entries generated from your environment. Expect more names than you explicitly installed because the packages also need dependencies.

::: {.callout-note collapse="true" title="Check your understanding"}
**Does `pip freeze` install packages?** No. It reports what is installed; `pip install -r requirements.txt` performs installation.

**What if you export while the terminal uses a different project's Python?** You record that other environment's packages. Check `sys.executable`, switch to the intended environment, and export again.

**What if you install another package afterward?** Repeat the import check and export again when you want the requirements file to reflect the updated environment.
:::

For details, see [pip's requirements-file guide](https://pip.pypa.io/en/stable/user_guide/#requirements-files).


## Find Local Data with Paths

A **path** describes where a file or folder is located. There are two ways to write one:

| Path type | Example | Where it starts |
|---|---|---|
| **Absolute** | `/Users/student/stat303-setup/data/station_readings.csv` on macOS, or `C:/Users/student/stat303-setup/data/station_readings.csv` on Windows | The filesystem root, including the drive on Windows |
| **Relative** | `data/station_readings.csv` | The current working directory of the program using the path |

The absolute examples include a particular user's folders, so they will usually differ on your computer. The relative example works when the notebook's current working directory is `stat303-setup` and the supplied folder layout is preserved.

The notebook and terminal can have different working directories. The folder visible in VS Code's Explorer does not prove either one's current directory. As a result, the terminal may find or create a file in one folder while the notebook looks for it in another. Check the current directory in each place when working with files.

Use Python's built-in `pathlib` module to build paths. With `Path`, `/` joins folder and file names and handles the separators for your operating system. In a **notebook code cell**:

```python
from pathlib import Path
print(Path.cwd())
data_path = Path("data") / "station_readings.csv"
print("Resolved path:", data_path.resolve())
print("File found:", data_path.is_file())
```

For the practice notebooks stored directly inside `stat303-setup`, the expected working directory is the extracted `stat303-setup` folder. If `is_file()` returns `False`, inspect the displayed directory, filename, capitalization, and extracted folder layout. VS Code settings or a different notebook location can change the working directory; always check rather than assume. If the directory is not the one you need, see *Move to the Folder You Need* below.

Once the file check returns `True`, inspect a few lines without needing data-analysis syntax:

```python
print(data_path.read_text(encoding="utf-8").splitlines()[:3])
```

For this supplied file, `read_text()` reads the text, `splitlines()` separates it into lines, and `[:3]` selects the first three. The result is a list of text lines, not yet a Pandas table.

A path expression describes a location; it does not establish that a file exists. In particular, `resolve()` displays an absolute path; it does not find a misplaced file or prove that the file exists.

### What Changes When the Working Directory Changes?

Suppose a different project has this layout. This is a hypothetical example; you do not need to create a `notebooks` folder in the practice kit.

If the notebook's current working directory is `example-project/notebooks`, `Path("data")` refers to a folder inside `notebooks`. To reach the CSV shown above, start with `..` to go up to `example-project`, then enter `data`:

This works only when the actual working directory has that relationship. Do not repeatedly add `..` until a path happens to work.

For the main workflow, keep a consistent folder layout and check `Path.cwd()` before building your data path.

### Move to the Folder You Need

When `Path.cwd()` is not the folder you expected, you have two ways to correct it: move yourself to where the notebook is, or move the running notebook's working directory to where you need it. Prefer the first.

**Preferred: open the right project folder.** A notebook's working directory normally starts as the folder that contains the notebook file. So an unexpected `Path.cwd()` usually means the notebook is not where you assumed. Extracting the practice kit can leave a nested `stat303-setup/stat303-setup/` folder, for example. In VS Code, choose **File → Open Folder** and open the `stat303-setup` folder that directly contains `activity02.ipynb` and the `data` folder, then reopen the notebook. This fix is permanent, requires no extra code, and gives the same working directory every time you open the project.

**When the files must stay where they are: `%cd`.** `%cd` is a line magic (Section 2.8) that changes the working directory of the running kernel, so every cell you run afterwards resolves relative paths from the new location.

```python
%cd data
```

Give `%cd` a child folder's name to move down, `..` to move up one level, or an absolute path to jump directly:

```python
%cd ..
%cd /Users/student/stat303-setup
%cd C:/Users/student/stat303-setup
```

Then confirm the result instead of assuming the move worked:

```python
%pwd
```

Two cautions apply. The change lasts only for the current kernel session, so restarting the kernel returns you to the starting directory. And `%cd` moves relative to where you already are, so running a `%cd ..` cell a second time takes you up two levels. If you use `%cd`, place it in one cell near the top of the notebook so that running the notebook from the top always produces the same working directory.

Finally, a terminal's `cd` and a notebook's `%cd` are separate. Typing `cd` in a terminal changes that terminal's directory and has no effect on a notebook that is already running, and `%cd` changes the kernel's directory and has no effect on your terminal.

::: {.callout-note collapse="true" title="Optional troubleshooting: other path checks"}
`!cd` runs in a temporary shell and does not change the notebook's working directory; use `%cd` as described above. Avoid scattering directory changes through your notebook, because they make results depend on the order in which cells run.

`data_path.exists()` checks whether the path exists, whether it is a file or a folder. For this activity, use `data_path.is_file()` because you are looking specifically for a file.
:::
