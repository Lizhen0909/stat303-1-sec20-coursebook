```python
import numpy as np
import pandas as pd
from pathlib import Path

# ---- messy input file for Part A -------------------------------------------
INVENTORY_RAW = """item,quantity,cost,restocked
Bolts,120,$0.45,2026-01-05
Nuts,300,$0.10,2026-01-05
Washers,unknown,$0.05,2026-01-06
Screws,210,$0.25,2026-01-06
Anchors,,$1.20,2026-01-07
Rivets,75,$0.30,2026-01-07
Clamps,40,$2.50,2026-01-08
Hinges,18,$3.75,2026-01-08
"""
Path('inventory_raw.csv').write_text(INVENTORY_RAW)

# ---- large table for the Part B timings ------------------------------------
rng = np.random.default_rng(303)
n = 200_000
orders = pd.DataFrame({
    'warehouse': rng.choice(['W1', 'W2', 'W3', 'W4'], size=n),
    'quantity': rng.integers(1, 60, size=n),
    'cost': rng.choice([3.00, 5.00, 8.00, 12.00, 18.00, 25.00], size=n),
})

# ---- small labeled table for Parts C and D ---------------------------------
item_columns = ['Bolts', 'Nuts', 'Washers', 'Screws']
warehouse = pd.DataFrame({
    'region':  ['Midwest', 'Midwest', 'East', 'West'],
    'Bolts':   [120,  85, 140,  60],
    'Nuts':    [300, 220, 260, 180],
    'Washers': [450, 380, 410, 275],
    'Screws':  [210, 165, 190, 130],
}, index=['W1', 'W2', 'W3', 'W4'])

unit_costs = np.array([0.45, 0.10, 0.05, 0.25])   # aligned to item_columns

print('NumPy version  :', np.__version__)
print('pandas version :', pd.__version__)
print('inventory_raw.csv written:', Path('inventory_raw.csv').exists())
print('orders shape   :', orders.shape)
print('unit_costs     :', unit_costs, '-> order:', item_columns)
warehouse
```
