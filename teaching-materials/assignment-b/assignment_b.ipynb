```python
from pathlib import Path
import pandas as pd

data = pd.read_csv(Path("data") / "library_visits.csv")
display(data)
print("Number of records:", len(data))
print("Mean visits:", data["visits"].mean())
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.bar(data["branch_id"], data["visits"])
ax.set_title("Illustrative Library Visits for One Day")
ax.set_xlabel("Branch ID")
ax.set_ylabel("Number of visits")
fig.tight_layout()
plt.show()
```
